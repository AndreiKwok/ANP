### Notebook onde é feito a ultima etapa de tratamento do dataset ANP + vars exogenas

Obs. as features: IPCA, SELIC & PIB_mensal são mensais

```python
import pandas as pd
```

```python
df_dolar_enriched = pd.read_parquet(r"..\data\dolar_enriched.parquet")
```

### Breve Tratamento feito no DataFrame ANP

```python
df_anp = pd.read_parquet("../data/download_gov/anp.parquet")
```

```python
df_anp.index = df_anp["Data da Coleta"]
```

```python
df_anp[(df_anp.index.year == 2026) & (df_anp.index.month == 4)]
```

### Merge ANP data with DataFrame dolar_enriched

```python
anp_enriched = pd.merge(
    df_dolar_enriched,
    df_anp, 
    left_index=True,
    right_index=True,
    how="inner", 
    )
```

```python
anp_enriched.to_parquet(r"..\data\anp_enriched.parquet")
```
