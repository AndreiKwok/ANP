## Consolidação — Modelos baseados em árvore (XGBoost e LightGBM)

### Experimentos realizados

Foram testadas 8 combinações independentes (2 algoritmos × 2 estratégias de
segmentação geográfica × 4 produtos), todas seguindo o mesmo protocolo:
tuning bayesiano via Optuna (`TimeSeriesSplit`, 8 folds), treino no bloco de
treino (2016-01 a 2025-04), avaliação em validação (2025-05 a 2025-08) e
teste final (2025-09 a 2026-08), sempre comparadas contra o baseline de
persistência (`lag_1`) no mesmo recorte.

### Segmentação por cluster estatístico

Como alternativa à segmentação por estado individual (27 dummies), foi
testada uma segmentação por **cluster de similaridade de preço**, definida
via Kruskal-Wallis (confirmação de diferença entre estados) + post-hoc de
Dunn (matriz de p-values par a par) + clustering hierárquico (average
linkage) sobre essa matriz — um cluster por produto, já que a estrutura de
similaridade entre estados variou conforme o produto (ex.: Diesel
apresentou 22 dos 26 estados testados em um único cluster, evidenciando
baixa variação regional; Gasolina apresentou 4 clusters de tamanho mais
distribuído).

### Resultado consolidado — MAE no teste final

| Produto  | Baseline | Estado + XGB | Cluster + XGB | Estado + LGBM | Cluster + LGBM |
|----------|----------|--------------|----------------|----------------|-----------------|
| Gasolina | 0.0597   | 0.0668       | 0.0676         | 0.0696         | 0.0748          |
| Etanol   | 0.0552   | 0.0612       | 0.0623         | 0.0619         | 0.0612          |
| Diesel   | 0.0815   | 0.1079       | 0.0994         | 0.0941         | **0.0908**      |
| GLP      | 1.0851   | 1.1281       | 1.1199         | 1.1441         | 1.1556          |

**Em nenhuma das 8 combinações o modelo superou ou igualou o baseline.**
O cenário mais próximo foi Diesel + cluster + LightGBM (0.0908 vs. 0.0815 —
ainda 11,4% pior).

### O que a clusterização revelou

A segmentação por cluster **melhorou** o resultado de forma consistente nos
dois algoritmos apenas para o **Diesel** (produto com maior homogeneidade
regional, coerente com sua forte atrelagem à paridade internacional de
preços). Para **Gasolina**, piorou de forma consistente nos dois algoritmos
— produto com heterogeneidade regional genuína, onde comprimir estados em
clusters descarta sinal informativo. Para Etanol e GLP, o efeito foi
inconsistente entre algoritmos, sugerindo impacto marginal, dentro da
margem de variação do tuning.

### Conclusão desta etapa

Oito tentativas independentes — variando algoritmo e granularidade
geográfica — convergiram para o mesmo resultado: nenhum modelo baseado em
árvore de decisão supera a simples persistência do último preço conhecido
no período de teste. Combinada com o diagnóstico anterior (viés sistemático
de previsão associado à mudança de patamar do Brent entre treino e teste,
fora do range observado), essa consistência reforça que a limitação é
estrutural à classe de modelos baseados em árvore — não uma peculiaridade
de implementação, hiperparâmetro ou segmentação testada.
